```python
import pandas as pd
df = pd.read_csv(r'C:\Users\Suzal\sales_data.csv')
df.head()
```

```python
# 1. See the summary of your data (Total rows, missing values, etc.)
print("--- Data Info ---")
print(df.info())

# 2. Get the basic math (Mean, Max, Min of your sales)
print("\n--- Basic Statistics ---")
print(df.describe())

# 3. Quick Insight: Which column has the highest values?
# Replace 'Sales' with your actual column name if it's different
if 'Sales' in df.columns:
    print("\n--- Total Revenue ---")
    print(df['Sales_Amount'].sum())
```

```python
# Which Region brings in the most money?
region_sales = df.groupby('Region')['Sales_Amount'].sum().sort_values(ascending=False)

print("--- Total Sales by Region ---")
print(region_sales)

# Who are the top 5 Sales Reps?
top_reps = df.groupby('Sales_Rep')['Sales_Amount'].sum().head(5)

print("\n--- Top 5 Sales Representatives ---")
print(top_reps)
```

```python
import matplotlib.pyplot as plt

# Create the bar chart
region_sales.plot(kind='bar', color='skyblue', edgecolor='black')

# Add labels and title
plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Sales Amount')
plt.xticks(rotation=45)

# Show the plot
plt.show()
```
